```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("to_vizu.tsv", sep="\t", index_col=None)

# Used cols by violin plot
# WARN: We use "orientation='h'" -> inverted ???
x_col="methyl"
y_col="gene"

df = df.head()
```

```python
# Violin plot
#box=True, points="all",
#
fig = px.violin(
        df, y=y_col, x=x_col,
        points="all",
        hover_data=df.columns,
        height=7500,
)
fig.update_traces(
        orientation='h',
)
fig.show()
```
